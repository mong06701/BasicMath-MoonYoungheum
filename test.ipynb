```python
from collections import deque

def solution(n, connections, starts):
    # 1. 인접 리스트 생성
    graph = []

    for _ in range(n + 1):
        graph.append([])

    for a, b in connections:
        graph[a].append(b)
        graph[b].append(a)

    # 2. BFS 준비
    queue = deque()
    visited = [False] * (n + 1)

    # 3. 시작 서버 등록
    for server in starts:
        queue.append((server, 0))
        visited[server] = True

    reached_count = 0
    last_minute = 0

    # 4. BFS 탐색
    while queue:
        current, minute = queue.popleft()

        reached_count += 1
        last_minute = minute

        for next_server in graph[current]:
            if not visited[next_server]:
                visited[next_server] = True
                queue.append((next_server, minute + 1))

    # 5. 모든 서버에 전달되었는지 확인
    if reached_count != n:
        return -1

    return last_minute
    
print(solution(
    6,
    [[1,3],[3,4],[2,5],[5,4],[4,6]],
    [1,2]
))
```

```python
import numpy as np

X = np.array([
    [4, 2],
    [1, 5]
])

w = np.array([2, -1])
b = 3

# 1. 기존 점수
print("기존 점수:", X @ w + b)
# [9 0]

# 2. 가중치 변경
w_new = np.array([2, -2])

print("가중치 변경:", X @ w_new + b)
# [ 7 -5]

# 3. 편향 변경
b_new = 5

print("편향 변경:", X @ w + b_new)
# [11  2]

# 4. 0벡터 입력
zero = np.array([0, 0])

print("0벡터 출력:", zero @ w + b)
# 3

# 5. 여러 출력값을 가지는 레이어
W = np.zeros((4, 3))
bias = np.zeros(3)

print("W Shape:", W.shape)
print("b Shape:", bias.shape)
print("전체 파라미터:", W.size + bias.size)

# W Shape: (4, 3)
# b Shape: (3,)
# 전체 파라미터: 15
```
