## 먼저 읽는 용어와 기호

| 용어·기호 | 쉬운 뜻 |

| 변수 / 상수 | 값을 바꿀 수 있는 문자 / 지금 계산에서 고정된 값 |
| 대입 | 식의 문자 자리에 주어진 숫자를 넣는 일 |
| 제곱 x² / 코드 `x ** 2` | x를 두 번 곱함. 3²=3×3=9 |

| 기울기 | 입력이 아주 조금 변할 때 출력이 어느 방향으로 얼마나 민감하게 변하는지 나타내는 값 |

| 손실 L | 예측과 정답의 차이를 평가한 값. 이 자료는 차이를 제곱한 손실을 사용 |
| 편미분 ∂L/∂w | 다른 변수는 고정하고 w만 아주 조금 바꿀 때의 순간 변화율 |
| Gradient ∇L | 여러 편미분값을 정해진 순서대로 묶은 벡터. ∇는 ‘나블라’로 읽음 |
| 벡터 [a,b] | 순서가 있는 숫자 묶음. 이 자료의 Gradient 순서는 항상 [w 방향, b 방향] |
| −∇L | Gradient의 각 성분에 −1을 곱해 방향을 뒤집은 벡터 |
| ≈ | 정확히 같은 값은 아니지만 대략 같다는 표시 |

**처음 보는 수식 읽기:** $\frac{f(x+h)-f(x)}{h}$에서 가로선 위가 분자, 아래가 분모입니다. 먼저 두 함수값을 빼고 그 결과를 h로 나눕니다. `∂L/∂w`는 이번 실습에서 L을 w로 단순히 나누라는 뜻이 아니라 **w에 대한 편미분을 나타내는 기호**입니다.

### 이번 실습에 사용하는 미분 규칙

| 함수의 항 | 미분한 결과 | 읽는 방법 |
|---|---|---|
| 상수 c | 0 | 입력이 달라도 상수는 변하지 않음 |
| ax | a | x의 계수가 기울기 |
| ax² | 2ax | 제곱항의 계수에 2를 곱하고 x를 남김 |
| 여러 항의 합 | 각 항을 미분한 결과의 합 | 항마다 따로 구한 뒤 더함 |

### 문제 1-1. 평균 변화율에서 미분값으로 연결하기

#### 문제 설명

함수는 $f(x)=x^2$입니다. 입력 x를 제곱한 값이 출력입니다. 먼저 구간 전체의 평균 변화율을 구하고, x=3에서 입력을 조금씩 바꿔 봅니다.

#### 요구사항

1. 구간 1→2, 2→3, 3→4, 4→5의 시작 함수값, 끝 함수값, 평균 변화율을 구하세요.
2. x=3으로 두고 h=1, 0.1, 0.01, 0.001일 때 아래 식을 계산하세요.
3. 결과가 어떤 수에 가까워지는지 확인하고 Q1~Q4에 답하세요.

$$\text{평균 변화율}=\frac{f(\text{끝 입력})-f(\text{시작 입력})}{\text{끝 입력}-\text{시작 입력}}$$

$$\frac{f(3+h)-f(3)}{h}$$

#### 작성 도움말

```python
def f(x):
    return x ** 2

starts = np.array([1, 2, 3, 4])
ends = np.array([2, 3, 4, 5])
rates = (f(ends) - f(starts)) / (ends - starts)
```

h는 `np.array([1, 0.1, 0.01, 0.001])`로 만들 수 있습니다. 교안의 미분 결과는 $f′(x)=2x$입니다. 이 식을 사용해 x=3에서의 미분값도 확인하세요. h 자리에 0을 직접 넣지 않습니다.

#### 해석 질문

**Q1.** 네 구간의 평균 변화율은 각각 얼마인가요?

**Q2.** h=1, 0.1, 0.01, 0.001일 때 변화율은 얼마이며 어떤 수에 가까워지나요?

**Q3.** f(3)과 f′(3)은 각각 얼마이며 뜻이 어떻게 다른가요?

**Q4.** 평균 변화율 식의 h에 0을 직접 넣어도 되나요?

#### 제출 결과

- 요구사항의 계산 결과
- Q1~Q4 답변

````python
# 문제 1-1. 평균 변화율에서 미분값으로 연결하기
# 1. 문제의 숫자와 제공된 식을 확인하세요.
# 2. 도움말을 참고해 요구사항 순서대로 계산하고 출력하세요.
# 3. 아래 답변란에 계산 결과와 뜻을 적으세요.
def f(x):
    return x ** 2

# 구간별 평균 변화율
starts = np.array([1, 2, 3, 4])
ends = np.array([2, 3, 4, 5])
rates = (f(ends) - f(starts)) / (ends - starts)

print("시작 함수값:", f(starts))
print("끝 함수값:", f(ends))
print("평균 변화율:", rates)

# x=3에서 h 줄여보기
h = np.array([1, 0.1, 0.01, 0.001])
rate_h = (f(3 + h) - f(3)) / h
for i in range(len(h)):
    print("h =", h[i], "-> 변화율:", round(rate_h[i], 4))

# 미분식 f'(x) = 2x
def f_prime(x):
    return 2 * x

print("f(3) =", f(3))
print("f'(3) =", f_prime(3))
````

#### 답변 작성란

**Q1.** 네 구간의 평균 변화율은 각각 얼마인가요?

답변: 1→2는 (4−1)/1=3, 2→3은 (9−4)/1=5, 3→4는 (16−9)/1=7, 4→5는 (25−16)/1=9이다. 뒤 구간으로 갈수록 평균 변화율이 커져서 x가 클수록 x²가 더 빨리 증가한다는 걸 알 수 있다.

**Q2.** h=1, 0.1, 0.01, 0.001일 때 변화율은 얼마이며 어떤 수에 가까워지나요?

답변: h=1이면 7, h=0.1이면 6.1, h=0.01이면 6.01, h=0.001이면 6.001이 나왔다. h를 줄일수록 6에 가까워지고, 이게 f′(x)=2x에 x=3을 넣은 값 6이랑 같다.

**Q3.** f(3)과 f′(3)은 각각 얼마이며 뜻이 어떻게 다른가요?

답변: f(3)=9이고 f′(3)=6이다. f(3)은 x=3을 넣었을 때 나오는 함수값(출력 자체)이고, f′(3)은 x=3에서 x를 아주 조금 바꿨을 때 출력이 얼마나 빠르게 변하는지(순간 변화율)를 나타내는 값이라서 의미가 다르다.

**Q4.** 평균 변화율 식의 h에 0을 직접 넣어도 되나요?

답변: 안 된다. h=0을 넣으면 분자는 f(3)−f(3)=0, 분모도 0이 돼서 0/0이 되므로 계산이 안 된다. 그래서 h를 0에 가깝게 점점 줄여가면서 어떤 값에 가까워지는지를 보는 것이다.

## 필수 실습 2. 편미분과 Gradient

### 문제 2-1. 한 변수만 움직이는 편미분 이해하기

#### 문제 설명

모델은 $\hat y=wx+b$, 입력 x=2, 실제 정답 y=10입니다. 현재 w=3, b=1입니다.

$$L(w,b)=(10-2w-b)^2$$
$$L(w,b)=4w^2+4wb+b^2-40w-20b+100$$
$$\frac{\partial L}{\partial w}=8w+4b-40,\qquad
\frac{\partial L}{\partial b}=4w+2b-20$$

#### 요구사항

1. 현재 예측값과 손실값을 계산하세요.
2. 제공된 식으로 w에 대한 편미분값과 b에 대한 편미분값을 구하세요.
3. **원래 점 (3,1)에서 각각 출발해** (3.1,1), (3,1.1)의 손실을 계산하세요. 앞의 변경을 다음 경우에 이어서 적용하지 않습니다.
4. 각 손실에서 원래 손실을 빼서 실제 변화량을 구하고 Q1~Q4에 답하세요.

#### 작성 도움말

```python
def loss(w, b):
    return (10 - 2 * w - b) ** 2
```

w만 바꾸는 경우는 `loss(3.1, 1)`, b만 바꾸는 경우는 `loss(3, 1.1)`입니다. 변화량은 **바꾼 뒤 손실 − 원래 손실**로 계산합니다. 편미분 식은 위 식을 그대로 사용해도 됩니다.

#### 해석 질문

````python
# 문제 2-1. 한 변수만 움직이는 편미분 이해하기
# 1. 문제의 숫자와 제공된 식을 확인하세요.
# 2. 도움말을 참고해 요구사항 순서대로 계산하고 출력하세요.
# 3. 아래 답변란에 계산 결과와 뜻을 적으세요.
def loss(w, b):
    return (10 - 2 * w - b) ** 2

w, b = 3, 1
x = 2
pred = w * x + b
L0 = loss(w, b)
print("예측값:", pred)
print("손실값:", L0)

# 편미분
dL_dw = 8 * w + 4 * b - 40
dL_db = 4 * w + 2 * b - 20
print("dL/dw:", dL_dw)
print("dL/db:", dL_db)

# 원래 점에서 하나씩만 0.1 증가
L_w = loss(3.1, 1)
L_b = loss(3, 1.1)
print("w만 0.1 증가 -> 손실:", round(L_w, 4), "변화량:", round(L_w - L0, 4))
print("b만 0.1 증가 -> 손실:", round(L_b, 4), "변화량:", round(L_b - L0, 4))
````

#### 답변 작성란

**Q1.** 원래 점 (w,b)=(3,1)의 예측값과 손실값은 얼마인가요?

답변: 예측값은 2×3+1=7이고, 손실값은 (10−7)²=9이다.

**Q2.** w와 b에 대한 편미분값은 각각 얼마이며 무엇을 고정하나요?

답변: ∂L/∂w=8×3+4×1−40=−12이고 이때는 b를 1로 고정하고 w만 움직인다. ∂L/∂b=4×3+2×1−20=−6이고 이때는 w를 3으로 고정하고 b만 움직인다.

**Q3.** w만 0.1 늘린 경우와 b만 0.1 늘린 경우의 손실 및 변화량은 각각 얼마인가요?

답변: w만 0.1 늘린 (3.1,1)은 손실 7.84, 변화량은 7.84−9=−1.16이다. b만 0.1 늘린 (3,1.1)은 손실 8.41, 변화량은 8.41−9=−0.59이다. 둘 다 손실이 줄었는데 w 쪽이 더 많이 줄었다(편미분값 크기가 w가 더 커서 그런 것 같다. −12×0.1=−1.2, −6×0.1=−0.6이랑 비슷함).

**Q4.** w에 대한 편미분값 −12를 “현재 손실이 −12”라고 읽어도 되나요?

답변: 안 된다. 현재 손실은 9이다. −12는 손실값이 아니라 b를 고정하고 w를 아주 조금 늘릴 때 손실이 줄어드는 정도(기울기)를 뜻한다. 손실은 제곱이라 음수가 될 수도 없다.

- 요구사항의 계산 결과

````python
# 문제 2-2. 편미분값을 묶어 Gradient 만들기
# 1. 문제의 숫자와 제공된 식을 확인하세요.
# 2. 도움말을 참고해 요구사항 순서대로 계산하고 출력하세요.
# 3. 아래 답변란에 계산 결과와 뜻을 적으세요.
gradient = np.array([-12, -6])   # [w 방향, b 방향]
negative_gradient = -gradient
print("Gradient:", gradient)
print("-Gradient:", negative_gradient)

L0 = loss(3, 1)
L_up = loss(3.1, 1.1)
L_down = loss(2.9, 0.9)
print("원래 손실:", L0)
print("(3.1, 1.1) 손실:", round(L_up, 4), "변화량:", round(L_up - L0, 4))
print("(2.9, 0.9) 손실:", round(L_down, 4), "변화량:", round(L_down - L0, 4))
````

**Q2.** −∇L은 무엇이며 Gradient와 어떤 관계인가요?

답변: −∇L은 [12, 6]이다. Gradient의 각 성분에 −1을 곱해서 방향을 반대로 뒤집은 벡터이고, Gradient가 손실이 커지는 쪽이라면 −∇L은 손실이 줄어드는 쪽 방향이다.

**Q3.** 두 변수를 모두 0.1 늘리거나 줄였을 때 손실과 변화량은 각각 얼마인가요?

답변: 둘 다 0.1 늘린 (3.1,1.1)은 손실 7.29, 변화량 −1.71로 줄었다. 둘 다 0.1 줄인 (2.9,0.9)는 손실 10.89, 변화량 +1.89로 오히려 늘었다.

**Q4.** 두 변수를 0.1씩 늘리는 이동을 정확한 −Gradient 방향이라고 부를 수 있나요?

### 문제 3-1. 다항식의 미분과 값 대입 복습하기

#### 제출 결과

- 요구사항의 계산 결과

````python
# 문제 3-1. 다항식의 미분과 값 대입 복습하기
# 1. 문제의 숫자와 제공된 식을 확인하세요.
# 2. 도움말을 참고해 요구사항 순서대로 계산하고 출력하세요.
# 3. 아래 답변란에 계산 결과와 뜻을 적으세요.
# C'(x) = 4x - 4  (2x^2 -> 4x, -4x -> -4, 3 -> 0)
def C(x):
    return 2 * x ** 2 - 4 * x + 3

def C_prime(x):
    return 4 * x - 4

x_values = np.array([1, 2, 3])
print("x:", x_values)
print("C(x):", C(x_values))
print("C'(x):", C_prime(x_values))
````

#### 답변 작성란

**Q1.** C′(x)는 무엇이며 각 항은 어떻게 미분되나요?

답변: C′(x)=4x−4이다. 2x²는 2×2x=4x, −4x는 계수인 −4, 상수 3은 변하지 않으니까 0이 되고 이걸 다 더하면 4x−4가 된다.

**Q2.** x=1,2,3의 함수값 C(x)는 각각 얼마인가요?

답변: C(1)=2−4+3=1, C(2)=8−8+3=3, C(3)=18−12+3=9이다.

**Q3.** x=1,2,3의 미분값 C′(x)는 각각 얼마인가요?

답변: C′(1)=0, C′(2)=4, C′(3)=8이다.

**Q4.** C′(1)=0이면 C(1)도 0인가요?

답변: 아니다. C′(1)=0은 x=1에서 기울기(순간 변화율)가 0이라는 뜻이고, 함수값은 원래 식에 넣어야 해서 C(1)=1이다. 미분값과 함수값은 다른 식에 대입한 값이다.

### 문제 3-2. 주어진 편미분식으로 Gradient 복습하기

#### 문제 설명

교안 연습문제 3의 모델을 사용합니다.

$$\hat y=wx+2bx,\qquad x=1,\ y=8,\ w=2,\ b=1$$

**이 문제의 b는 일반적인 ŷ=wx+b의 편향과 다릅니다.** 교안에 주어진 식 그대로 2bx의 계수입니다. x=1이므로 예측식은 w+2b가 됩니다.

$$L(w,b)=(8-w-2b)^2$$
$$L(w,b)=w^2+4wb+4b^2-16w-32b+64$$
$$\frac{\partial L}{\partial w}=2w+4b-16,\qquad
\frac{\partial L}{\partial b}=4w+8b-32$$

#### 요구사항

1. 현재 예측값과 손실값을 구하세요.
2. 제공된 두 편미분식에 w=2, b=1을 대입하세요.
3. [w 방향, b 방향] 순서의 Gradient와 −Gradient를 만드세요.
4. Q1~Q4에 답하세요. 벡터의 크기, 단위 방향, 내적 근사, 안장점 계산은 요구하지 않습니다.

#### 작성 도움말

예측은 `w * x + 2 * b * x`, 손실은 `(y - prediction) ** 2`입니다. 두 편미분값을 `np.array([partial_w, partial_b])`로 묶고 앞에 `-`를 붙이면 반대 방향입니다. 전개식과 미분식은 위 식을 참고하세요.

#### 해석 질문

**Q1.** 현재 예측값과 손실값은 얼마인가요?

**Q2.** w와 b에 대한 편미분값은 각각 얼마인가요?

**Q3.** Gradient와 −Gradient는 각각 무엇인가요?

**Q4.** 이번 모델을 앞 문제처럼 ŷ=2w+b로 계산해도 되나요?

#### 제출 결과

- 요구사항의 계산 결과
- Q1~Q4 답변

````python
# 문제 3-2. 주어진 편미분식으로 Gradient 복습하기
# 1. 문제의 숫자와 제공된 식을 확인하세요.
# 2. 도움말을 참고해 요구사항 순서대로 계산하고 출력하세요.
# 3. 아래 답변란에 계산 결과와 뜻을 적으세요.
x, y = 1, 8
w, b = 2, 1

prediction = w * x + 2 * b * x
loss_val = (y - prediction) ** 2
print("예측값:", prediction)
print("손실값:", loss_val)

partial_w = 2 * w + 4 * b - 16
partial_b = 4 * w + 8 * b - 32
print("dL/dw:", partial_w)
print("dL/db:", partial_b)

grad = np.array([partial_w, partial_b])
print("Gradient:", grad)
print("-Gradient:", -grad)
````

#### 답변 작성란

**Q1.** 현재 예측값과 손실값은 얼마인가요?

답변: 예측값은 2×1+2×1×1=4이고 손실값은 (8−4)²=16이다.

**Q2.** w와 b에 대한 편미분값은 각각 얼마인가요?

답변: ∂L/∂w=2×2+4×1−16=−8, ∂L/∂b=4×2+8×1−32=−16이다.

**Q3.** Gradient와 −Gradient는 각각 무엇인가요?

답변: Gradient는 [w 방향, b 방향] 순서로 [−8, −16]이고 −Gradient는 [8, 16]이다.

**Q4.** 이번 모델을 앞 문제처럼 ŷ=2w+b로 계산해도 되나요?

답변: 안 된다. 이번 모델은 ŷ=wx+2bx이고 x=1이라 ŷ=w+2b이다. 2w+b로 하면 4+1=5가 나와서 실제 예측값 4와 달라지고, 손실이랑 편미분도 다 틀리게 된다. b가 일반적인 편향이 아니라 2bx의 계수라는 점을 주의해야 한다.
