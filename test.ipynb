```python
from collections import deque

def solution(n, connections, starts):
    # 1. 인접 리스트 생성
    graph = []

    for _ in range(n + 1):
        graph.append([])

    for a, b in connections:
        graph[a].append(b)
        graph[b].append(a)

    # 2. BFS 준비
    queue = deque()
    visited = [False] * (n + 1)

    # 3. 시작 서버 등록
    for server in starts:
        queue.append((server, 0))
        visited[server] = True

    reached_count = 0
    last_minute = 0

    # 4. BFS 탐색
    while queue:
        current, minute = queue.popleft()

        reached_count += 1
        last_minute = minute

        for next_server in graph[current]:
            if not visited[next_server]:
                visited[next_server] = True
                queue.append((next_server, minute + 1))

    # 5. 모든 서버에 전달되었는지 확인
    if reached_count != n:
        return -1

    return last_minute
    
print(solution(
    6,
    [[1,3],[3,4],[2,5],[5,4],[4,6]],
    [1,2]
))
```

```python
def solution(orders, queries):
    # 코드를 작성하세요.
    result = []

    for target in queries:
        left = 0
        right = len(orders) - 1

        found = -1
        while left <= right:
            mid = (left + right) // 2
            if orders[mid] == target:
                found = mid
                break
            elif orders[mid] < target:
                left = mid + 1
            else: 
                right = mid - 1

        result.append(found)
    return result

print(solution([1, 2, 3, 4, 5], [3, 6, 1]))
```

```python
import numpy as np

# 입력 데이터와 실제 정답
x = 2.0
y = 10.0

# 초기 파라미터
w = 3.0
b = 1.0

# 학습률
learning_rate = 0.05

# 예측 함수
def predict(x, w, b):
    return w * x + b

# 손실 함수
def loss_fn(x, y, w, b):
    y_pred = predict(x, w, b)
    return (y - y_pred) ** 2

# Gradient 계산
def gradient_fn(x, y, w, b):
    y_pred = predict(x, w, b)

    grad_w = 2 * (y_pred - y) * x
    grad_b = 2 * (y_pred - y)

    return np.array([grad_w, grad_b])

# 파라미터를 벡터로 표현
theta = np.array([w, b])

# 반복 학습
for step in range(6):
    w, b = theta

    # 1. 순전파
    y_pred = predict(x, w, b)

    # 2. 손실 계산
    loss = loss_fn(x, y, w, b)

    # 3. Gradient 계산
    grad = gradient_fn(x, y, w, b)

    print(
        f"Step {step}: "
        f"w={w:.4f}, "
        f"b={b:.4f}, "
        f"Loss={loss:.4f}"
    )

    # 4. 파라미터 업데이트
    theta = theta - learning_rate * grad

# Step 0: w=3.0000, b=1.0000, Loss=9.0000
# Step 1: w=3.6000, b=1.3000, Loss=2.2500
# Step 2: w=3.9000, b=1.4500, Loss=0.5625
# Step 3: w=4.0500, b=1.5250, Loss=0.1406
# Step 4: w=4.1250, b=1.5625, Loss=0.0352
# Step 5: w=4.1625, b=1.5813, Loss=0.0088
```
