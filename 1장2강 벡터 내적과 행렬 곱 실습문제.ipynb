### 문제 1-1. 내적으로 고객별 가중합 계산하기

#### 문제 설명

교안의 고객 활동 데이터와 가중치입니다.

| 고객 | 방문 횟수 | 장바구니 추가 횟수 |
|---|---:|---:|
| A | 3 | 2 |
| B | 5 | 1 |
| C | 2 | 4 |
| 가중치 | 2 | 4 |

입력과 가중치의 순서는 **방문 횟수 → 장바구니 추가 횟수**입니다. 이번 문제는 편향을 더하지 않고 내적만 계산합니다.

$$x\cdot w=x_1w_1+x_2w_2$$

- $x_1$, $x_2$: 고객의 첫 번째·두 번째 입력값
- $w_1$, $w_2$: 같은 위치의 특성에 적용할 가중치
- $\cdot$: 같은 위치끼리 곱한 값을 모두 더하는 내적 기호

#### 수식 읽는 순서

`x₁w₁`은 ‘첫 입력값 × 첫 가중치’입니다. 두 글자를 붙여 썼지만 곱셈을 뜻합니다. `x₂w₂`도 같은 방식입니다. `x·w`는 이 곱셈 결과들을 모두 더한 값입니다.

‘대응하는 원소’는 **같은 순서에 있는 숫자끼리**라는 뜻입니다. 방문 횟수에는 방문 가중치를, 장바구니 횟수에는 장바구니 가중치를 곱하세요.

#### 요구사항

#### 제출 결과

- 고객별 입력·가중치와 Shape
- 세 고객의 점수 및 np.dot·@ 비교 결과
- 대입 계산 과정과 Q1~Q5 답변

````python
# 문제 1-1. 내적으로 고객별 가중합 계산하기
# 요구사항을 한 단계씩 작성하고 실행하세요.
# 계산 방법과 함수 이름은 위의 설명을 참고하세요.

# 1. 가중치와 고객별 입력 (방문 -> 장바구니 순서)
w = np.array([2, 4])
x_a = np.array([3, 2])
x_b = np.array([5, 1])
x_c = np.array([2, 4])
print("w:", w)
print("x_a:", x_a, "/ x_b:", x_b, "/ x_c:", x_c)

# 2. Shape 확인
print("x_a.shape:", x_a.shape, "/ w.shape:", w.shape)

# 3. np.dot으로 고객별 점수
score_a = np.dot(x_a, w)
score_b = np.dot(x_b, w)
score_c = np.dot(x_c, w)
print("A 점수:", score_a)
print("B 점수:", score_b)
print("C 점수:", score_c)

# 4. @ 연산자로도 계산해서 비교
print("x_a @ w:", x_a @ w)
print("np.dot과 같은가?", score_a == x_a @ w)
````

#### 답변 작성란

**수식 및 계산 과정**

- 작성: x·w = x₁w₁ + x₂w₂ 에 대입. A: 3×2 + 2×4 = 6 + 8 = 14 / B: 5×2 + 1×4 = 10 + 4 = 14 / C: 2×2 + 4×4 = 4 + 16 = 20. x_a.shape와 w.shape가 둘 다 (2,)로 원소 개수가 같아서 내적이 가능하다.

**Q2. 내적 결과는 벡터인가요, 스칼라인가요? `np.dot(x_a, w)`와 `x_a @ w`의 결과는 같은가요?**

- 답변: 내적 결과는 숫자 하나인 스칼라이다. np.dot(x_a, w)와 x_a @ w 모두 14로 결과가 같았다.

**Q3. A와 B의 입력값은 다른데 점수가 같을 수 있나요? 이번 계산을 근거로 설명하세요.**

**Q4. 원소가 2개인 입력과 원소가 3개인 가중치로 일반적인 벡터 내적을 계산할 수 있나요?**

### 문제 1-2. 내적의 부호와 코사인 유사도 이해하기

#### 문제 설명

교안의 간단한 방향 예시를 사용합니다. 아래의 모든 벡터는 0벡터가 아닙니다.

| 구분 | 벡터 |
|---|---|
| 기준 벡터 `a` | [1, 0] |
| 같은 방향 `b_same` | [2, 0] |
| 수직 방향 `b_perp` | [0, 2] |
| 반대 방향 `b_opposite` | [−2, 0] |

추가로 교안의 `u = [3, 4]`, `v = [6, 8]`의 코사인 유사도를 계산합니다.

$$\mathrm{cosine\ similarity}=\frac{u\cdot v}{\|u\|_2\times\|v\|_2}$$

분자는 두 벡터의 내적, 분모는 두 벡터의 L2 Norm을 곱한 값입니다.

| 결과 | 방향 관계 |
|---|---|
| 코사인 유사도 1 | 같은 방향 |
| 코사인 유사도 0 | 서로 수직 |
| 코사인 유사도 −1 | 반대 방향 |

> 교안의 코사인 유사도 목록 중 ‘반대 방향: 1’은 ‘반대 방향: −1’의 부호 오타이므로 바로잡아 적용했습니다.

#### 먼저 이해하기

- 벡터를 원점에서 출발하는 화살표로 생각하면, ‘방향’은 화살표가 향하는 쪽입니다.
- `[1,0]`과 `[2,0]`은 같은 쪽, `[1,0]`과 `[0,2]`는 직각, `[1,0]`과 `[−2,0]`은 서로 반대쪽을 향합니다.
- 90°는 직각, 180°는 완전히 반대 방향입니다. 각도를 구하는 코드는 작성하지 않고 표와 내적의 부호를 연결합니다.
- **분자**는 분수의 위쪽, **분모**는 아래쪽입니다. 코사인 유사도 식은 ‘내적 ÷ (첫 길이 × 둘째 길이)’로 읽습니다.
- L2 길이는 이전 강의처럼 ‘각 원소 제곱 → 더하기 → 제곱근’으로 구합니다. 예를 들어 √25는 5×5=25이므로 5입니다.

1~2번은 방향별 내적, 3~4번은 u·v의 코사인 유사도입니다. 두 부분을 차례로 진행하세요.

````python
# 문제 1-2. 내적의 부호와 코사인 유사도 이해하기
# 요구사항을 한 단계씩 작성하고 실행하세요.
# 계산 방법과 함수 이름은 위의 설명을 참고하세요.

# 1. 방향 비교용 벡터
a = np.array([1, 0])
b_same = np.array([2, 0])
b_perp = np.array([0, 2])
b_opposite = np.array([-2, 0])

# 2. a와 각 벡터의 내적
print("a · b_same:", np.dot(a, b_same))
print("a · b_perp:", np.dot(a, b_perp))
print("a · b_opposite:", np.dot(a, b_opposite))

# 3. u, v의 내적과 L2 Norm
u = np.array([3, 4])
v = np.array([6, 8])
dot_uv = np.dot(u, v)
norm_u = np.linalg.norm(u)
norm_v = np.linalg.norm(v)
print("dot_uv:", dot_uv)
print("norm_u:", norm_u)
print("norm_v:", norm_v)

# 4. 코사인 유사도 = 내적 / (길이1 * 길이2)
cos_sim = dot_uv / (norm_u * norm_v)jnn
print("코사인 유사도:", cos_sim)
````

#### 답변 작성란

**수식 및 계산 과정**

- 작성: a·b_same = 1×2 + 0×0 = 2 / a·b_perp = 1×0 + 0×2 = 0 / a·b_opposite = 1×(−2) + 0×0 = −2. u·v = 3×6 + 4×8 = 18 + 32 = 50, ‖u‖₂ = √(9+16) = 5, ‖v‖₂ = √(36+64) = 10, 코사인 유사도 = 50 / (5×10) = 50 / 50 = 1.

**Q3. u와 v의 내적, 두 L2 Norm, 코사인 유사도를 순서대로 계산하세요.**

행렬 두 개의 곱은 다음 규칙을 따릅니다.

$$(m,n)\ @\ (n,p)\ \longrightarrow\ (m,p)$$

m은 고객 수, n은 입력 특성 수, p는 출력 수로 읽을 수 있습니다. 가운데 크기가 같아야 계산할 수 있습니다.

두 행렬의 Shape는 `(행 수, 열 수)`입니다. `(m,n) @ (n,p)`에서 먼저 **왼쪽 열 수와 오른쪽 행 수**를 비교합니다. 이 둘이 같으면 한 행과 한 열의 숫자를 빠짐없이 짝지어 내적할 수 있습니다.

결과의 행 수는 왼쪽 행렬에서, 결과의 열 수는 오른쪽 행렬에서 가져옵니다. 이 문제에서는 **고객 수 × 출력 점수 종류 수**입니다.

1~2번은 고객별 한 점수, 3~4번은 고객별 두 점수, 5번은 Shape를 읽는 연습입니다. 5번은 실제 배열을 만들지 않고 숫자의 개수만 비교하세요.

````python
# 문제 2-2. 여러 고객의 한 개·두 개 점수와 Shape 확인
# 요구사항을 한 단계씩 작성하고 실행하세요.
# 계산 방법과 함수 이름은 위의 설명을 참고하세요.

# 1. X, w, w_col, W 만들기
X = np.array([
    [3, 2],
    [5, 1],
    [2, 4]
])
w = np.array([2, 4])
w_col = np.array([
    [2],
    [4]
])
W = np.array([
    [2, 1],
    [4, 3]
])
print("X.shape:", X.shape)
print("w.shape:", w.shape)
print("w_col.shape:", w_col.shape)
print("W.shape:", W.shape)

# 2. 고객별 점수 하나
scores = X @ w
scores_col = X @ w_col
print("scores:", scores, "/ shape:", scores.shape)
print("scores_col:")
print(scores_col)
print("scores_col.shape:", scores_col.shape)

# 3. 고객별 점수 두 개
Y = X @ W
print("Y:")
print(Y)
print("Y.shape:", Y.shape)
````

**Q3. 고객 A의 두 점수는 얼마인가요? W의 어느 열과 계산했는지도 설명하세요.**

#### 제출 결과

- X·W·Y의 Shape와 Y 출력
- 네 원소의 손계산 과정

````python
# 문제 3-2. 두 고객의 작은 행렬곱 복습
# 요구사항을 한 단계씩 작성하고 실행하세요.
# 계산 방법과 함수 이름은 위의 설명을 참고하세요.

# 1. X, W 만들고 Shape 확인
X = np.array([
    [2, 3],
    [4, 1]
])
W = np.array([
    [1, 2],
    [3, 1]
])
print("X.shape:", X.shape, "/ W.shape:", W.shape)

# 2. 행렬곱
Y = X @ W
print("Y:")
print(Y)
print("Y.shape:", Y.shape)

# 4. 원소별 곱셈과 비교
print("X * W:")
print(X * W)
print("첫 행 첫 열 -> X @ W:", Y[0, 0], "/ X * W:", (X * W)[0, 0])
````

#### 답변 작성란

**수식 및 계산 과정**

- 작성: Y11 = X 1행 [2,3] · W 1열 [1,3] = 2×1 + 3×3 = 2 + 9 = 11 / Y12 = X 1행 [2,3] · W 2열 [2,1] = 2×2 + 3×1 = 4 + 3 = 7 / Y21 = X 2행 [4,1] · W 1열 [1,3] = 4×1 + 1×3 = 4 + 3 = 7 / Y22 = X 2행 [4,1] · W 2열 [2,1] = 4×2 + 1×1 = 8 + 1 = 9. 따라서 Y = [[11, 7], [7, 9]]. X * W = [[2×1, 3×2], [4×3, 1×1]] = [[2, 6], [12, 1]].

**Q1. X와 W의 Shape는 무엇이며 행렬곱이 가능한 이유는 무엇인가요?**

- 답변: X와 W 모두 Shape가 (2, 2)이다. X의 열 수 2와 W의 행 수 2가 같아서 행렬곱이 가능하다.

**Q2. 고객 A의 두 점수는 각각 얼마인가요? 선택한 행·열과 계산 과정을 적으세요.**

- 답변: X의 1행 [2, 3]을 고르고, 첫 점수(구매 관심도)는 W의 1열 [1, 3]과 2×1 + 3×3 = 11, 둘째 점수(고객 활동)는 W의 2열 [2, 1]과 2×2 + 3×1 = 7이다.

**Q3. 고객 B의 두 점수와 Y의 Shape는 무엇인가요? 결과의 행과 열이 무엇을 뜻하는지 설명하세요.**

- 답변: B는 X의 2행 [4, 1]을 써서 첫 점수 4×1 + 1×3 = 7, 둘째 점수 4×2 + 1×1 = 9이다. Y의 Shape는 (2, 2)이고, 행은 고객(A, B), 열은 출력 점수 종류(구매 관심도, 고객 활동)를 뜻한다.

**Q4. X @ W와 X * W의 첫 행 첫 열 값은 각각 얼마인가요? 왜 다른가요?**

- 답변: X @ W의 첫 행 첫 열은 11, X * W는 2이다. X @ W는 X의 1행과 W의 1열을 내적(2×1 + 3×3)하고, X * W는 같은 위치 원소끼리만 곱(2×1)하기 때문에 다르다.
