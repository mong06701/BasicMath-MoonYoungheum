## 필수 1. 선형·아핀 변환과 파라미터 이해하기

### 문제 1-1. 선형 변환에 편향을 더해 보기

#### 문제 설명

교안 1절의 예제입니다. 선형 변환은 입력에 가중치 행렬을 곱하는 계산이며, 아핀 변환은 그 결과에 편향을 더하는 계산입니다.

```python
x = np.array([3, 2])
W = np.array([[2, 0], [0, 3]])
bias = np.array([1, -2])
zero = np.array([0, 0])
```

- 선형 변환: `x @ W`
- 아핀 변환: `x @ W + bias`

W의 첫 번째 열은 [2,0], 두 번째 열은 [0,3]입니다. 0벡터는 모든 값이 0인 벡터를 말합니다.

#### 요구사항

1. x, W, bias, zero를 위와 같이 만드세요.
2. `linear_output = x @ W`, `affine_output = x @ W + bias`를 계산하여 출력하세요.
3. `zero @ W`와 `zero @ W + bias`를 출력하세요.
4. 답변 작성란에 두 출력값의 곱셈·덧셈 과정을 적으세요.

#### 작성 도움말

첫 출력은 x의 [3,2]와 W의 첫 열 [2,0]을 같은 순서로 짝지어 `3×2 + 2×0`으로 계산합니다. 둘째 출력도 둘째 열을 사용해 같은 방법으로 계산하세요. 편향은 첫 출력에 1, 둘째 출력에 −2를 각각 더합니다.

#### 해석 질문

**Q1.** 선형 변환의 첫 번째·두 번째 출력은 각각 어떻게 계산되나요?

**Q2.** 편향을 더한 아핀 변환 결과는 무엇인가요?

**Q3.** 0벡터 입력에서 두 변환의 결과는 각각 무엇인가요?

**Q4.** 편향이 [0,0]이라면 두 변환의 결과가 같아지나요?

#### 제출 결과

- 두 변환의 출력과 0벡터의 출력
- 계산 과정과 Q1~Q4 답변

````python
# 문제 1-1. 선형 변환에 편향을 더해 보기
# 위의 데이터와 작성 도움말을 보고 한 단계씩 코드를 작성하세요.
# 중간 결과를 출력하여 확인하세요.

# 1. 입력, 가중치, 편향, 0벡터 만들기
x = np.array([3,2])
W = np.array([[2,0], [0,3]])
bias = np.array([1,-2])
zero = np.array([0,0])
# 2. 선형 변환, 아핀 변환
linear_output = x @ W
affine_output = x @ W + bias
print("linear_output:", linear_output)
print("affine_output:", affine_output)
# 3. 0벡터 넣어보기
print("zero @ W:", zero @ W)
print("zero @ W + bias:", zero @ W + bias)
````

#### 답변 작성란

**수식 및 계산 과정**

- 작성: 선형 변환 첫 출력 = x [3,2]와 W 첫 열 [2,0]: 3×2 + 2×0 = 6, 둘째 출력 = x [3,2]와 W 둘째 열 [0,3]: 3×0 + 2×3 = 6 → [6, 6]. 아핀 변환은 여기에 편향을 더해서 [6+1, 6+(−2)] = [7, 4]. 0벡터는 0×2 + 0×0 = 0, 0×0 + 0×3 = 0이라서 선형 변환은 [0, 0], 아핀 변환은 [0+1, 0+(−2)] = [1, −2].

**Q1. 선형 변환의 첫 번째·두 번째 출력은 각각 어떻게 계산되나요?**

- 답변: 첫 번째 출력은 x와 W의 첫 열 [2, 0]을 곱해서 3×2 + 2×0 = 6, 두 번째 출력은 W의 둘째 열 [0, 3]과 곱해서 3×0 + 2×3 = 6이다. 그래서 선형 변환 결과는 [6, 6]이다.

**Q2. 편향을 더한 아핀 변환 결과는 무엇인가요?**

- 답변: 선형 변환 결과 [6, 6]에 편향 [1, −2]를 위치별로 더해서 [7, 4]이다.

**Q3. 0벡터 입력에서 두 변환의 결과는 각각 무엇인가요?**

- 답변: 선형 변환은 [0, 0]이고, 아핀 변환은 [1, −2]이다. 입력이 0이면 곱한 값이 다 0이 돼서 선형 변환은 0벡터가 나오지만, 아핀 변환은 편향이 그대로 남는다.

**Q4. 편향이 [0,0]이라면 두 변환의 결과가 같아지나요?**

- 답변: 같아진다. 편향이 [0, 0]이면 더해도 값이 안 바뀌니까 x @ W + bias = x @ W가 된다.

### 문제 1-2. 가중치와 편향을 하나씩 바꾸기

#### 문제 설명

교안 연습문제 2의 데이터입니다. 각 고객의 첫 입력은 방문 횟수, 둘째 입력은 장바구니 추가 횟수입니다.

```python
X = np.array([[4, 2], [1, 5]])
w = np.array([2, -1])
b = 3
```

X의 첫 행 [4,2]는 고객 A, 둘째 행 [1,5]는 고객 B입니다. 고객별 점수는 ‘첫 입력×첫 가중치 + 둘째 입력×둘째 가중치 + 편향’으로 계산합니다.

| 비교 | 가중치 | 편향 |
|---|---|---|
| 기본 | [2,−1] | 3 |
| 가중치만 변경 | [2,−2] | 3 |
| 편향만 변경 | [2,−1] | 5 |

두 변경은 각각 기본 설정에서 출발합니다.

#### 요구사항

1. 위 배열을 만들고 `base_scores = X @ w + b`를 계산하여 출력하세요.
2. `w_new = np.array([2, -2])`로 만들고 `weight_scores = X @ w_new + b`를 계산하세요. 점수와 `weight_scores - base_scores`를 출력하세요.
3. 기본 w를 사용하여 `b_new = 5`, `bias_scores = X @ w + b_new`를 계산하세요. 점수와 `bias_scores - base_scores`를 출력하세요.
4. `zero = np.array([0, 0])`를 만들고 기본 설정인 `zero @ w + b`를 출력하세요.

#### 작성 도움말

변화량은 ‘변경 후 점수 − 기본 점수’입니다. 변화량이 −2라면 점수가 2만큼 줄었다는 뜻입니다. 편향만 바꿀 때는 w_new가 아닌 기본 w를 사용하세요. 한 번에 한 조건만 바꿉니다.

#### 해석 질문

**Q1.** 기본 설정에서 고객 A와 B의 점수는 각각 얼마인가요?

**Q2.** 가중치만 바꾸면 각 점수와 변화량은 얼마인가요?

**Q3.** 편향만 바꾸면 각 점수와 변화량은 얼마인가요?

**Q4.** 기본 설정에 0벡터를 넣으면 점수가 얼마인가요? 왜 그 값이 남나요?

#### 제출 결과

- 기본 점수, 두 변경의 점수와 변화량, 0벡터의 점수
- 계산 과정과 Q1~Q4 답변

````python
# 문제 1-2. 가중치와 편향을 하나씩 바꾸기
# 위의 데이터와 작성 도움말을 보고 한 단계씩 코드를 작성하세요.
# 중간 결과를 출력하여 확인하세요.

# 1. 기본 설정
X = np.array([[4, 2], [1, 5]])
w = np.array([2, -1])
b = 3
base_scores = X @ w + b
print("base_scores:", base_scores)

# 2. 가중치만 변경
w_new = np.array([2, -2])
weight_scores = X @ w_new + b
print("weight_scores:", weight_scores)
print("변화량:", weight_scores - base_scores)

# 3. 편향만 변경 (가중치는 기본 w 사용)
b_new = 5
bias_scores = X @ w + b_new
print("bias_scores:", bias_scores)
print("변화량:", bias_scores - base_scores)

# 4. 0벡터 입력
zero = np.array([0, 0])
print("zero @ w + b:", zero @ w + b)
````

#### 답변 작성란

**수식 및 계산 과정**

- 작성: 기본: A = 4×2 + 2×(−1) + 3 = 8 − 2 + 3 = 9, B = 1×2 + 5×(−1) + 3 = 2 − 5 + 3 = 0. 가중치만 변경 [2, −2]: A = 8 − 4 + 3 = 7, B = 2 − 10 + 3 = −5. 편향만 변경 5: A = 8 − 2 + 5 = 11, B = 2 − 5 + 5 = 2. 0벡터: 0×2 + 0×(−1) + 3 = 3.

**Q1. 기본 설정에서 고객 A와 B의 점수는 각각 얼마인가요?**

- 답변: 고객 A는 4×2 + 2×(−1) + 3 = 9, 고객 B는 1×2 + 5×(−1) + 3 = 0이다.

**Q2. 가중치만 바꾸면 각 점수와 변화량은 얼마인가요?**

- 답변: 점수는 A = 7, B = −5이고 변화량은 [7−9, −5−0] = [−2, −5]이다. 둘째 가중치가 −1에서 −2로 바뀌어서 장바구니 횟수만큼 점수가 줄었다. A는 장바구니가 2라서 2만큼, B는 5라서 5만큼 줄었다. 그래서 고객마다 변화량이 다르다.

**Q3. 편향만 바꾸면 각 점수와 변화량은 얼마인가요?**

- 답변: 점수는 A = 11, B = 2이고 변화량은 [2, 2]이다. 편향은 입력과 상관없이 마지막에 더하는 값이라 3에서 5로 2만큼 늘리면 모든 고객 점수가 똑같이 2씩 늘어난다.

**Q4. 기본 설정에 0벡터를 넣으면 점수가 얼마인가요? 왜 그 값이 남나요?**

- 답변: 0×2 + 0×(−1) + 3 = 3이다. 입력이 0이면 가중치를 곱한 부분은 전부 0이 되고, 마지막에 더하는 편향 3만 남기 때문이다.

## 필수 2. ReLU와 신경망 순전파

### 문제 2-1. ReLU와 한 레이어를 순서대로 계산하기

#### 문제 설명

교안 3절의 예제입니다. 먼저 ReLU의 규칙을 연습한 뒤 레이어 하나를 계산합니다.

**ReLU(렐루)의 규칙:** 양수는 그대로 두고, 음수는 0으로 바꿉니다. 0은 0으로 둡니다.

```python
def relu(z):
    return np.maximum(0, z)

test_z = np.array([-3, -1, 0, 2, 5])
X = np.array([[3, 2], [1, 4], [0, 2]])
W = np.array([[1, -2], [2, 1]])
b = np.array([-4, 1])
```

X의 세 행은 고객 A, B, C입니다. W의 두 열은 뉴런 두 개의 가중치입니다. 첫 뉴런의 가중치는 [1,2], 둘째 뉴런의 가중치는 [−2,1]입니다. 편향은 각각 −4와 1입니다.

이번 계산은 **가중치 적용 → 편향 덧셈 → ReLU**의 세 단계입니다.

#### 요구사항

1. 위 코드를 입력하고 `relu(test_z)`를 출력하세요.
2. `weighted = X @ W`를 계산하여 출력하세요.
3. `Z = weighted + b`를 계산하여 출력하세요.
4. `A = relu(Z)`를 계산하여 출력하세요.
5. `Z.shape`와 `A.shape`를 출력하고, 고객 A의 두 뉴런 계산을 손으로 확인하세요.

#### 작성 도움말

`def relu(z):`와 그 아래 `return` 줄은 위 예제를 그대로 사용하세요. return 앞의 들여쓰기도 유지하세요. 고객 A의 첫 뉴런은 `3×1 + 2×2`를 구한 뒤 −4를 더합니다. 마지막에 그 값과 0 중 큰 값을 고릅니다.

#### 해석 질문

**Q1.** ReLU는 [−3,−1,0,2,5]를 어떻게 바꾸나요?

**Q2.** 고객 A의 가중합, 편향을 더한 값 Z, ReLU 결과 A는 각각 무엇인가요?

**Q3.** 고객 B와 C의 Z와 ReLU 결과는 각각 무엇인가요?

#### 제출 결과

- ReLU 연습 결과, weighted·Z·A의 출력, Z·A의 Shape
- 고객 A의 계산 과정과 Q1~Q4 답변

````python
# 문제 2-1. ReLU와 한 레이어를 순서대로 계산하기
# 위의 데이터와 작성 도움말을 보고 한 단계씩 코드를 작성하세요.
# 중간 결과를 출력하여 확인하세요.

# 1. ReLU 함수와 데이터
def relu(z):
    return np.maximum(0, z)

test_z = np.array([-3, -1, 0, 2, 5])
X = np.array([[3, 2], [1, 4], [0, 2]])
W = np.array([[1, -2], [2, 1]])
b = np.array([-4, 1])
print("relu(test_z):", relu(test_z))

# 2. 가중치 적용
weighted = X @ W
print("weighted:")
print(weighted)

# 3. 편향 더하기
Z = weighted + b
print("Z:")
print(Z)

# 4. ReLU 적용
A = relu(Z)
print("A:")
print(A)

# 5. Shape 확인
print("Z.shape:", Z.shape, "/ A.shape:", A.shape)
````

**Q2. 고객 A의 가중합, 편향을 더한 값 Z, ReLU 결과 A는 각각 무엇인가요?**

- 답변: 가중합은 [7, −4] (3×1 + 2×2 = 7, 3×(−2) + 2×1 = −4), 편향을 더한 Z는 [7−4, −4+1] = [3, −3], ReLU 결과 A는 [3, 0]이다.

**Q3. 고객 B와 C의 Z와 ReLU 결과는 각각 무엇인가요?**

````python
# 문제 2-2. 은닉층의 결과를 출력층에 전달하기
# 위의 데이터와 작성 도움말을 보고 한 단계씩 코드를 작성하세요.
# 중간 결과를 출력하여 확인하세요.

# 1. ReLU 함수와 입력, 파라미터
def relu(z):
    return np.maximum(0, z)

X = np.array([[3, 2], [1, 4], [0, 2]])
W1 = np.array([[1, -2], [2, 1]])
b1 = np.array([-4, 1])
W2 = np.array([[2], [-1]])
b2 = np.array([1])

# 2. 순서대로 계산 (은닉층 -> ReLU -> 출력층)
Z1 = X @ W1 + b1
H = relu(Z1)
Y = H @ W2 + b2
print("Z1:")
print(Z1)
print("H:")
print(H)
print("Y:")
print(Y)

# 3. Shape 확인
print("H.shape:", H.shape, "/ Y.shape:", Y.shape)
````

**Q2. 고객 A·B·C의 최종 점수는 각각 얼마인가요? 계산 과정을 적으세요.**

- 답변: A = 3×2 + 0×(−1) + 1 = 7, B = 5×2 + 3×(−1) + 1 = 8, C = 0×2 + 3×(−1) + 1 = −2이다.

**Q3. H의 Shape (3,2)와 Y의 Shape (3,1)는 각각 무엇을 뜻하나요?**

- 답변: H의 (3, 2)는 고객 3명 × 은닉층 뉴런 2개의 출력이고, Y의 (3, 1)은 고객 3명 × 최종 점수 1개라는 뜻이다.

**Q4. 고객 C의 최종 점수가 음수로 나와도 되는 이유는 무엇인가요?**

### 실습 마무리 답변 작성란

**1. 선형 변환과 아핀 변환의 계산은 어떻게 다른가요?**

- 답변: 선형 변환은 입력에 가중치 행렬을 곱하기만 하고(xW), 아핀 변환은 그 결과에 편향까지 더한다(xW + b). 그래서 0벡터를 넣으면 선형 변환은 0이 나오고 아핀 변환은 편향이 남는다.

**2. 가중치와 편향은 각각 언제 사용하는 숫자인가요?**

- 답변: 가중치는 입력값에 곱할 때 쓰는 숫자이고, 편향은 곱해서 다 더한 결과에 마지막으로 더할 때 쓰는 숫자이다.

**3. ReLU는 [−3,−1,0,2,5]를 어떻게 바꾸나요?**

- 답변: [0, 0, 0, 2, 5]로 바꾼다. 음수는 0이 되고 0과 양수는 그대로 둔다.

**4. 두 레이어의 계산에서 출력층에 전달하는 값은 무엇인가요?**

- 답변: 은닉층의 Z1에 ReLU를 적용한 H를 출력층에 전달한다.

**5. 최종 출력의 Shape (3,1)는 무엇을 뜻하나요?**

- 답변: 고객 3명 각각에 대해 최종 점수 1개씩 나왔다는 뜻이다. (행: 고객, 열: 출력 점수)

**6. 주어진 가중치와 편향으로 출력만 계산하면 학습이 이루어지나요?**

- 답변: 아니다. 주어진 가중치와 편향으로 출력만 계산한 것은 순전파일 뿐이고, 가중치와 편향 값을 바꾸지 않았으니 학습이 이루어진 게 아니다.
